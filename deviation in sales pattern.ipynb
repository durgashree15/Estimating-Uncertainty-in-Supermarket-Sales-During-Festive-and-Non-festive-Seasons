```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_excel("sales product-date.xlsx")
```

```python
col_names=list(data.columns)[1:]
```

```python
daily=[]
weekly=[]
std_based=[]
```

```python
for i in data.index:
    temp=[]
    for j in col_names:
        temp.append(data[j][i])
    temp.sort()
    temp=list(set(temp))
    q1, q3= np.percentile(temp,[25,75])
    iqr = q3 - q1
    lb = q1 -(1.5 * iqr) 
    ub = q3 +(1.5 * iqr)
    for j in col_names:
        if int(data[j][i])>ub:
            if j=="20-Oct" or j=="21-Oct" or j=="22-Oct" or j=="23-Oct" or j=="24-Oct" or j=="25-Oct" or j=="26-Oct" or j=="27-Oct" or j=="28-Oct":
                print(data['Row Labels'][i])
                daily.append(data['Row Labels'][i])
         
    for j in col_names:
        if int(data[j][i])<lb:
            if j=="20-Oct" or j=="21-Oct" or j=="22-Oct" or j=="23-Oct" or j=="24-Oct" or j=="25-Oct" or j=="26-Oct" or j=="27-Oct" or j=="28-Oct":
                print(data['Row Labels'][i])
```
